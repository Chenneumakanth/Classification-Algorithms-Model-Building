**Encode categorical variables**

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df["Category"] = le.fit_transform(df["Category"])
df["LaunchMonth"] = le.fit_transform(df["LaunchMonth"])
df["Country"] = le.fit_transform(df["Country"])
df["Currency"] = le.fit_transform(df["Currency"])
df["VideoIncluded"] = le.fit_transform(df["VideoIncluded"])
```

```python
#Feature target
y = df[['IsSuccessful']]
y
```

```python
#Feature selection
from sklearn.feature_selection import SelectKBest,f_classif
df_numeric = df.select_dtypes(include=['number'])
selector = SelectKBest(score_func=f_classif, k=8)


x= selector.fit_transform(df_numeric,y)
feature_scores = selector.scores_

selected_indices = selector.get_support(indices=True)
chosen_feature_names = df_numeric.columns[selected_indices]

# 2. Convert your numpy array 'x' back into a beautiful DataFrame
x = pd.DataFrame(x, columns=chosen_feature_names)
```

```python
##x = x.drop('RaisedAmount', axis=1, errors='ignore')
```

```python
x = x.drop('IsSuccessful', axis=1, errors='ignore')
```

```python
x
```

```python
import warnings
```

```python
from sklearn.model_selection import train_test_split,GridSearchCV


x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=16)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC

models = {
    "Random Forest": RandomForestClassifier(random_state=42),
    "KNN (K-Nearest Neighbors)": KNeighborsClassifier(n_neighbors=5),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Naive Bayes": GaussianNB(),
    "Support Vector Machine (SVM)": SVC(kernel='rbf', random_state=42)
}
results = {}
for name, model in models.items():
   model.fit(x_train, y_train)
```

```python

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
evaluation_results = []



for name, model in models.items():
    y_train_pred = model.predict(x_train)
    y_test_pred = model.predict(x_test)

    # 2. Calculate requested Performance Metrics
    metrics = {
        "Train Accuracy": accuracy_score(y_train, y_train_pred),
        "Test Accuracy": accuracy_score(y_test, y_test_pred),
        "Train Precision": precision_score(y_train, y_train_pred),
        "Test Precision": precision_score(y_test, y_test_pred),
        "Train Recall": recall_score(y_train, y_train_pred),
        "Test Recall": recall_score(y_test, y_test_pred),
        "Train F1-Score": f1_score(y_train, y_train_pred),
        "Test F1-Score": f1_score(y_test, y_test_pred),
    }
    evaluation_results.append(metrics)

    # 3. Generate Confusion Matrices
    cm_train = confusion_matrix(y_train, y_train_pred)
    cm_test = confusion_matrix(y_test, y_test_pred)

    # Print individual reports to console
    print(f"  {name} Metrics:")
    print(f"   [Train vs Test Accuracy]: {metrics['Train Accuracy']:.4f} vs {metrics['Test Accuracy']:.4f}")
    print(f"   [Train vs Test F1-Score]: {metrics['Train F1-Score']:.4f} vs {metrics['Test F1-Score']:.4f}")
    print(f"   Test Confusion Matrix:\n  {cm_test}\n")

```

Decision Tree achieved the highest test accuracy (99.31%), closely followed by the SVM (98.40%). While the Random Forest and Decision Tree still hit a perfect 1.0000 on the training data, they only dropped by ~1% on the test set. Naive Bayes (66.70%) and KNN (84.22%) performed poorly because they cannot handle columns that correlate with each other

**Apply Cross-Validation to evaluate model stability**

```python
from sklearn.model_selection import StratifiedKFold,GridSearchCV,RandomizedSearchCV
cv=StratifiedKFold(n_splits=5,shuffle=True,random_state=42)
```

```python
dt_param_grid = {
    'criterion': ['gini', 'entropy'],
    'max_depth': [None, 3, 5, 10, 15],
    'min_samples_split': [2, 5, 10]
}
```

```python
dt_grid = GridSearchCV(
    estimator=DecisionTreeClassifier(random_state=42),
    param_grid=dt_param_grid,
    cv=cv,
    scoring='accuracy',
    verbose=1,
    n_jobs=-1
)
dt_grid.fit(x_train, y_train)
dt_best_parameter = dt_grid.best_params_
print(f'Best parameters:{dt_best_parameter}')
dt_best_score = dt_grid.best_score_
print(f'Best score:{dt_best_score}')
```

##**Task 5 – Model Comparison & Selection**

####**Compare all classification models using evaluation metrics**

```python
best_models = {
    "Random Forest": rf_grid.best_estimator_,
    "KNN": knn_grid.best_estimator_,
    "Decision Tree": dt_grid.best_estimator_,
    "Naive Bayes": nb_grid.best_estimator_,
    ##"SVM": svc_grid.best_estimator_
}
final_comparison = []
for name, model in best_models.items():
    y_pred = model.predict(x_test)
    final_comparison.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred),
        "Recall": recall_score(y_test, y_pred),
        "F1-Score": f1_score(y_test, y_pred)
    })

df_comparison = pd.DataFrame(final_comparison).sort_values(by="F1-Score", ascending=False)
df_comparison
```

**Discuss strengths, weaknesses, and possible improvements**

• **Decision Tree**
	• Strength: Achieves peak accuracy (99.46%) by perfectly capturing crisp, razor-sharp logical boundaries in the data.
	• Weakness: Highly sensitive to data changes; risk of breaking if future data distributions drift.

• **Random Forest**
	• Strength: Excellent generalization and highly resilient against noise or dataset variance.
	• Weakness: Ensemble averaging slightly blurs the exact mathematical thresholds, dropping accuracy to 98.33%.

• **K-Nearest Neighbors (KNN)**
	• Strength: Simple, intuitive distance-based model that requires zero formal training time.
	• Weakness: Struggled heavily with multidimensional noise, leading to over 1,500 misclassifications (85.59% accuracy).

  
• **Naive Bayes**
	• Strength: Computationally lightweight, lightning-fast, and highly scalable.
	• Weakness: Total mathematical failure (67.05% accuracy); its assumption of independent features completely breaks down on correlating crowdfunding columns.
